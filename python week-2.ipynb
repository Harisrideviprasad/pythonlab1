```python
#finding greatest of 3 numbers
a=int(input("enter first value:"))
b=int(input("enter second value:"))
c=int(input("enter third value:"))
if(a>b and a>c):
  print(f"{a} is greater")
elif(b>a and b>c):
   print(f"{b} is greater")
else:
   print(f"{c} is greater")
```

```python
#program on "match"
num=int(input("enter any num:"))
match num:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
```
