```python
#print multiplication table of given number
n=int(input())
for i in range(1,21):
 print(f" {n} * {i} = {n*i}")
```

```python
#check given num is prime or not
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("it is not a prime number")
    break
  else:
    print("yes it is prime")
```

```python
#print prime numbers btwn 2 numbers
a=int(input())
b=int(input())
for i in range(a,b+1):
 for j in range(2,i):
  if(i%j==0):
    break
 else:
    print(i,end=" ")
```

```python
#print sum of digits of given number
n=int(input())
sum=0
while(n!=0):
  remainder = n%10
  sum = sum+remainder
  n = n//10
print(sum)
```
