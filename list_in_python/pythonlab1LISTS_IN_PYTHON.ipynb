```python
l1=[1,2,3,55,66,88,99,0,"hello","name"]
l2=[100,220,440]
print(l1+l2)#concatenation
print(l2*3)#repetition
print(l1[0])#indexing
print(22 in l1)#membership operator
print(l1[ : :-1])#print elements in reverse order
print(l1[2:7])#slicing,prints elements from second index to 6th index
print(l1[-7:-1])
for i in l2:
  print(i) #iteration through a list
```
