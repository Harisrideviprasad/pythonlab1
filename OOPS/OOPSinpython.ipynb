phython prog represents a class shape to calculate area and perimeter of shapes circle,square,triangle


```python
class shape:
 def area():
  pass
 def perimeter():
  pass
class circle(shape):
  def __init__(self,r):
    self.r=r
  def area(self):
    return 3.14*self.r*self.r
  def perimeter(self):
    return 2*3.14**self.r
class square(shape):
  def __init__(self,s):
    self.s=s
  def area(self):
    return self.s*self.s
  def perimeter(self):
    return 4*self.s
class triangle(shape):
  def __init__(self,a,b,c,h):
    self.a=a
    self.b=b
    self.c=c
    self.h=h
  def area(self):
    return 0.5*self.b*self.h
  def perimeter(self):
    return self.a+self.b+self.c
c1=circle(4)
print(c1.area())
print(c1.perimeter())
s1=square(5)
print(s1.area())
print(s1.perimeter())
t1=triangle(4,5,6,7)
print(t1.area())
print(t1.perimeter())
```
