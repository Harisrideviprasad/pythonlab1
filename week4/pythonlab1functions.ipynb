```python
#program to check whether given number is palindrome or not using functions
def palindrome(n):
  org=n
  rev=0
  while n>0:
    digit=n%10
    rev=rev*10+digit
    n=n//10
  if org==rev:
      return True
  else:
      return False
num=int(input())
if palindrome(num):
     print("palindrome")
else:
     print("not palindrome")
```

```python
#program to reverse a string using function
def reverse(a):
  rev=""
  for ch in a:
    rev=ch+rev
  return rev
srng=input()
print(reverse(srng))
```

```python
def reversestr(s):
  rev=s[ : :-1 ]
  return rev
srng=input()
print(reversestr(srng))
```

```python
#default argument
def print_msg(name,branch="AIDS"):
  print("the name is",name,"the branch is",branch)
print_msg("hari")
```

```python
#positional arg
def print_msg(name,branch):
  print("the name is",name,"the branch is",branch)
print_msg("AIDS","hari")
```

```python
#keyword arg
def print_msg(name,branch):
  print("the name is",name,"the branch is",branch)
print_msg(branch="AIDS",name="hari")
```

```python
#variable-length arg
def add(*numbers):
  print(sum(numbers))
add(1,2,3,4,5)
```

```python
#function that returns multiple values
def multiple_values(*num):
  s=sum(num)
  l=len(num)
  avg=s/l
  return s,l,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"sum is {res_sum},count is {res_count},avg is {res_avg}")
```
